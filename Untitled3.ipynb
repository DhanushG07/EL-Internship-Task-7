```python
import sqlite3

# Create a new SQLite database file in memory
conn = sqlite3.connect("sales_data.db")
cursor = conn.cursor()

# Create the sales table
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    product TEXT,
    quantity INTEGER,
    price REAL
)
''')

# Insert sample sales data
data = [
    ('Pen', 10, 1.5),
    ('Pencil', 20, 0.5),
    ('Notebook', 15, 2.0),
    ('Eraser', 25, 0.75),
    ('Marker', 12, 1.25),
]

cursor.executemany("INSERT INTO sales (product, quantity, price) VALUES (?, ?, ?)", data)
conn.commit()
print("Database and table created with sample data.")
```

```python
import pandas as pd

# SQL query
query = '''
SELECT product,
       SUM(quantity) AS total_qty,
       SUM(quantity * price) AS revenue
FROM sales
GROUP BY product
'''

# Load result into pandas
df = pd.read_sql_query(query, conn)
print(df)
```

```python
import matplotlib.pyplot as plt

# Plot revenue by product
df.plot(kind='bar', x='product', y='revenue', title='Revenue by Product', legend=False)
plt.ylabel('Revenue')
plt.tight_layout()
plt.savefig("sales_chart.png")
plt.show()
```
